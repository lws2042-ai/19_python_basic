```python
# while 반복문 사용하기
#초기식
#while 조건식:
#    반복할 코드
#    변화식

i = 0

while i < 100:
    print("hello world",i)
    i += 1
```

```python
# 초깃값을 1부터 시작하기

i = 1
while i <= 10:
    print("Hello world", i)
    i += 1
```

```python
# 초깃값을 감소시키기

i = 10
while i > 0:
    print("hello", i)
    i -= 1
```

```python
# 입력한 횟수대로 반복하기

count = int(input())

i = 0
while i < count:
    print("Hello world!", i)
    i += 1
```

```python
# 반복횟수가 정해지지 않는 경우

import random

i = 0
while i != 3:
    i = random.randint(1, 6)
    print(i)

#random.randint(1,6) # 1~6까지 랜덤 숫자를 줌
```
