```python
#변수만들기
# 영문문자와 숫자를 사용할 수 있음
# 대소문자를 구분합니다.
# 문자부터 시작해야 하며(숫자가 앞에 나오면 안됨)
# _(밑줄문자)로 시작할 수 없습니다.
# 특수문자는 사용할 수 없다.
# 파이썬의 키워드(IF, ELSE, WHILE 등)는 변수명으로 사용할 수 없다.
```

```python
x = 10
y = "hello world"
print(y)
```

```python
# 변수의 자료형 알아보기
print(type(x))
print(type(y))
```

```python
# 변수 여러개를 한 번에 만들기
x, y, z = 10, 20, 30
print(x)
print(y)
print(z)
```

```python
x = y = z = 10
print(x)
print(y)
print(z)
```

```python
x, y = 10, 20
x, y = y, x
print(x)
print(y)
```

```python
# 변수로 계산하기
a = 10
b = 20
c = a + b
print(c)
```

```python
# 산술 연산 후 할당 연산자 사용하기
a = 10
a += 20 # a = a + 20
print(a)
```

```python
# 입력 값을 변수에 저장하기

a = int(input("첫 번째 숫자를 입력하세요:"))
b = int(input("두 번째 숫자를 입력하세요:"))

print(a + b)
```

```python
# 입력 값을 변수 두 개에 저장하기
a, b = input("문자열 두 개를 입력하세요").split()
print(a)
print(b)
```
