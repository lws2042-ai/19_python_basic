```python
# 반복문으로 2차원 리스트의 요소를 모두 출력하기
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]
for x, y in a:
    print(x, y)
```

```python
# for 반복문을 두 번 사용하기
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]
for i in a:
    for j in i:
        print(j, end=" ")
    print()
```

```python
# for와 ragne 사용하기
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]
for i in range(len(a)):
    for j in range(len(a[i])):
        print(a[i][j], end=" ")
    print()
```

```python
# while 반복문을 한 번 사용하기

a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]

i = 0
while i < len(a):
    x, y = a[i]
    print(x, y)
    i += 1
```

```python
# 23.7 심사문제: 지뢰찾기 p285

import copy

input_grid = [] # 입력 받은 값을 저장하는 리스트이다.
result_grid = [] # 결과값을 저장하는 리스트이다.

# step1. 2차원 리스트의 행 개수와 열 개수를 입력 받는다.
row, col = map(int, input().split())   

 
# step2.  행 개수 만큼 값을 입력 받는다.
for r in range(row):
    input_grid.append(list(input()))

result_grid = copy.deepcopy(input_grid) # 출력용 리스트에 입력받은 리스트 값 복사. copy 사용시 같은 객체를 참조함. copy.deepcopy 사용함(chat GPT 참조)

# step3.  Target Cell의 주변 8개 값 검사 
for r in range(row):        # 행(row) 개수 만큼 검사한다.
    for c in range(col):    # 열(col) 개수 만큼 검사한다.

        #print(">>>", r, c, input_grid[r][c])  # Target Cell 

        if input_grid[r][c] == '*': # 지뢰는 count 제외
            continue

        hit_count = 0

        for nbr_row in range(r-1, r+2):
            for nbr_col in range(c-1, c+2):

                 # if (조건1) or (조건2) or (조건3) 설명 => 조건1: 자기 자신은 제외, 조건2: 존재하지 않는 행은 제외, 조건3: 존재하지 않는 열은 제외
                if (r == nbr_row and c == nbr_col) or (nbr_row < 0 or nbr_row >= row) or (nbr_col < 0 or nbr_col >= col):
                    continue 

                # '*'(지뢰의 수를 카운트 한다.)
                if input_grid[nbr_row][nbr_col] == '*':
                    hit_count += 1

                #print(nbr_row, nbr_col, input_grid[nbr_row][nbr_col])

        result_grid[r][c] = hit_count    

#print(result_grid)


# step4. 결과를 화면에 출력한다.
for output in result_grid:
    print(''.join(map(str, output)))    # map(str, output) 행의 모든 원소를 문자열로 변화 (chat GPT 참조)
        
```
