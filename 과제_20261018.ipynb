```python
# 문제 14. 성적 관리 프로그램

# 학생들의 성적을 관리하는 프로그램을 작성하시오.

# 다음 학생들의 성적 정보를 딕셔너리로 저장하시오
# 학생1: 이름=김민수, 국어=85, 영어=90, 수학=78
# 학생2: 이름=이영희, 국어=92, 영어=88, 수학=95
# 학생3: 이름=박철수, 국어=78, 영어=85, 수학=82

# 요구사항:
# 1. 각 학생의 총점과 평균을 계산하여 딕셔너리에 추가
# 2. 각 학생의 정보를 보기 좋게 출력


students_info = {
                    '학생1': {'이름': '김민수', '국어': 85, '영어': 90, '수학': 78},
                    '학생2': {'이름': '이영희', '국어': 92, '영어': 88, '수학': 95},
                    '학생3': {'이름': '박철수', '국어': 78, '영어': 85, '수학': 82}
}

print ("=== 학생 성적 관리 ===")

for stdts_key in students_info:
    #print(stdts_key, stdt_value)

    if stdts_key != '학생1':
        print("---")

    kor = students_info[stdts_key]['국어']
    eng = students_info[stdts_key]['영어']
    mat = students_info[stdts_key]['수학']

    tot = kor +  eng +  mat
    avg = round(tot / 3,2)

    print("이름:", students_info[stdts_key]['이름'])
    print("국어: ", kor,"점",", 영어: ", eng,"점", ", 수학: ", mat,"점", sep="" )
    print("총점: ", tot,"점",", 평균: ", avg,"점", sep="" )
    


#**출력 예시:**

#=== 학생 성적 관리 ===
#이름: 김민수
#국어: 85점, 영어: 90점, 수학: 78
#총점: 253점, 평균: 84.33점
#---
#이름: 이영희
#국어: 92점, 영어: 88점, 수학: 95점
#총점: 275점, 평균: 91.67점
#---
#이름: 박철수
#국어: 78점, 영어: 85점, 수학: 82점
#총점: 245점, 평균: 81.67점# 
```

```python
# 23.7 심사문제: 지뢰찾기 p285

input_grid = [] # 입력 받은 값을 저장하는 리스트이다.
result_grid = [] # 결과값을 저장하는 리스트이다.

# step1. 2차원 리스트의 행 개수와 열 개수를 입력 받는다.
row, col = map(int, input().split())  
 
# step2.  행 개수 만큼 값을 입력 받는다.
for r in range(row):
    input_grid.append(list(input()))

# 입력 리스트와 출력 리스트를 분리 한다.
result_grid = [line[:] for line in input_grid]

# step3.  Target Cell의 주변 8개 값 검사 
for r in range(row):        # 행(row) 개수 만큼 검사한다.
    for c in range(col):    # 열(col) 개수 만큼 검사한다.

        #print(">>>", r, c, input_grid[r][c])  # Target Cell 

        if input_grid[r][c] == '*': # 지뢰는 count 제외
            continue

        hit_count = 0

        for nbr_row in range(r-1, r+2):
            for nbr_col in range(c-1, c+2):

                 # if (조건1) or (조건2) or (조건3) 설명 => 조건1: 자기 자신은 제외, 조건2: 존재하지 않는 행은 제외, 조건3: 존재하지 않는 열은 제외
                if (r == nbr_row and c == nbr_col) or (nbr_row < 0 or nbr_row >= row) or (nbr_col < 0 or nbr_col >= col):
                    continue 

                # '*'(지뢰의 수를 카운트 한다.)
                if input_grid[nbr_row][nbr_col] == '*':
                    hit_count += 1

                #print(nbr_row, nbr_col, input_grid[nbr_row][nbr_col])

        result_grid[r][c] = hit_count    

#print(result_grid)


# step4. 결과를 화면에 출력한다.
for output in result_grid:
    print(''.join(map(str, output)))    # map(str, output) 행의 모든 원소를 문자열로 변화 (chat GPT 참조)
```
