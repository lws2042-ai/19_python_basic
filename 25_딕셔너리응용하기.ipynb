```python
# 딕셔너리에 키와 기본값 저장하기
x = {
    'a': 10,
    'b': 20,
    'c': 30,
    'd': 40
}

x.setdefault("e")
x.setdefault("f", 100)
x
```

```python
# 딕셔너리의 키값 수정하기
x = {
    'a': 10,
    'b': 20,
    'c': 30,
    'd': 40
}
x.update(a=90) # x["a"] = 90과 동일함
x
```

```python
# 딕셔너리에 키-값 쌍 삭제하기
x = {
    'a': 10,
    'b': 20,
    'c': 30,
    'd': 40
}
x.pop("a") 
x
```

```python
# 딕셔너리의 모든 키-값 쌍을 삭제하기
x = {
    'a': 10,
    'b': 20,
    'c': 30,
    'd': 40
}
x.clear() 
x
```

```python
# 딕셔너리의 키의 값을 가져오기
x = {
    'a': 10,
    'b': 20,
    'c': 30,
    'd': 40
}
print(x.get("a"))
print(x.get("z", 0))
```

```python
# 딕셔너리의 모든 키의 값을 가져오기
x = {
    'a': 10,
    'b': 20,
    'c': 30,
    'd': 40
}
for key, value in x.items():
    print(key, ': ', value, sep='')
```

```python
# 딕셔너리의 키만 출력하기
x = {
    'a': 10,
    'b': 20,
    'c': 30,
    'd': 40
}
for key in x.keys():
    print(key, end=' ')

```

```python
# 딕셔너리의 값만 출력하기

x = {
    'a': 10,
    'b': 20,
    'c': 30,
    'd': 40
}
for value in x.values():
    print(value, end=' ')
```

```python
# 딕셔너리 표현식 사용하기
keys = ['a', 'b', 'c', 'd']
x = {key: value for key, value in dict.fromkeys(keys).items()}
x
```
