```python
# while에서 break로 반복문 끝내기

i = 0
while True: #무한루프
    print(i)
    i += 1
    if i == 100: # i가 10일 때
        break   # 반복문을 끝냄. while의 제어흐름을 벗어남
```

```python
# for에서 break 반복문 끝내기

for i in range(100):
    print(i)
    if i == 10:
        break
```

```python
# continue로 코드 실행 건너뛰기
for i in range(100):
    if i % 2 == 0:  # 짝수이면 
        continue    # 이후 실행문들 건너뜀
    print(i)        # 홀수만 출력됨
```

```python
i = 0
while i < 100:
    i += 1
    if i % 2 == 0:
        continue
    print(i)
```

```python
# 입력한 횟수대로 반복하기

count = int(input("반복할 횟수를 입력하세요:"))

i = 0
while True:
    print(i)
    i += 1
    if i == count:
        break
```

```python
# 입력한 숫자까지 홀수 출력하기

count = int(input("반복할 횟수를 입력하세요:"))

for i in range(count + 1):
    if i % 2 == 0:
        continue
    print(i)
```
