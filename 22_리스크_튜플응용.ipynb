```python
# 반복문으로 리스트의 요소를 모두 출력하기

a = [38, 21,  53, 62, 19]

for i in enumerate(a):
    print(i)
```

```python
# 인덱스와 요소를 함께 출력하기

a = [38, 21,  53, 62, 19]
for index, value in enumerate(a):
    print(index, value)
```

```python
# 인덱스와 요소를 함께 출력하기
# for 인덱스, 요소 in enumerate(리스트):

a = [38, 21, 53, 62, 19]
for index, value in enumerate(a):
    print(index, value)
```

```python
a = [38, 21, 53, 62, 19]
for index, value in enumerate(a, start=1):
    print(index, value)
```

```python
# while 반복문으로 요소 출력하기

a = [38, 21, 53, 62, 19]

i = 0
while i < len(a):
    print(a[i])
    i += 1
```

```python
# 리스트의 가장 작은수 구하기

a = [38, 21, 53, 62, 19]
smallest = a[0]
for i in a:
    if i < smallest:
        smallest = i
print('smallest:',smallest)

# 리스트의 가장 큰수 구하기
a = [38, 21, 53, 62, 19]
largest = a[0]
for i in a:
    if i > largest:
        largest = i
print('largest:',largest)
```

```python
# 리스트 표현식 사용하기
a = [i for i in range(10)]  # 0부터 9까지 숫자를 생성하여 리스트 생성
print(a)

b = list(i for i in range(10))
print(b)
```

```python
# for 반복문과 if 조건문을 여러 번 사용하기
a = [i * j 
     for i in range(2, 10) 
     for j in range(1, 10)]
a
```

```python
c = []
for i in range(2, 10):
    for j in range(1, 10):
        c.append(i * j)
print(c)
```

```python
# 리스트에 map사용하기
#list(map(함수, 리스트))
#tuple(map(함수, 튜플))

a = [1.2, 2.5, 3.7, 4.6]

a = list(map(int, a))
a[0] = 0
print(a)

a = tuple(map(int, tuple(a)))
print(a)
```
