```python
# for 와 range 사용하기
for i in range(100):
    print("Hello, world!")
```

```python
# 반복문에서 변수의 변화 알아보기

for i in range(10):
    print("hello, world", i)
```

```python
# 시작하는 숫자와 끝나는 숫자 지정하기
for j in range(10, 20):
    print("hello world", j)
```

```python
# 증가폭 사용하기
for k in range(0, 10, 2):
    print("hello world", k)
```

```python
# 숫자를 감소 시키기
for i in range(10, 0, -1):
    print("hello world", i)
```

```python
a = [1, 3, 5, 9, 7]
print(list(a))
print(list(reversed(a)))
```

```python
for i in reversed(range(10)):
    print(i)
```

```python
# 시쿼스 객체로 반복하기
a = [10, 20, 30, 40, 50]

for num in a:
    print(num)
```

```python
fruits = ('apple','orange','grape')
for fruit in fruits:
    print(fruit)
```
