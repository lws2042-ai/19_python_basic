```python
# 함수의 호출 과정 알아보기
def mul(a, b):
    c = a * b
    return c

def add(a, b):
    c = a + b
    print(c)
    d = mul(a,b)
    print(d)

x = 10
y = 20
add(x, y)
```

```python
# 연습문제: 몫과 나머지를 구하는 함수 만들기 p384
x = 10
y = 3

def get_quotient_remainder(x, y):
    quo = x // y
    rem = x % y
    return quo, rem

quotient, remainder = get_quotient_remainder(10, 3)
print(f'몫: {quotient}, 나머지: {remainder}')
```

```python
# 29.8 심사문제: 사칙 연산 함수 만들기

x, y = map(int, input().split())

def calc(x, y):
    return x + y, x - y, x * y, x / y


a, s, m, d = calc(x, y)

print(f'덧셈: {a}, 뺄셈: {s}, 곱셈: {m}, 나눗셈: {d}')
```
