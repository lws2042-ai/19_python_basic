```python
# range()를 사용하여 리스트 만들기
range(0, 10) # 0~9까지의 정수를 꺼내 사용할수 있음
list(range(0,10))
```

```python
range(5, 12) # 5~11까지의 정수를 꺼내 사용할수 있음
b = list(range(5, 12))
print(b)
```

```python
c = list(range(0, 20, 2)) # 0~19까지의 정수 중에서 2씩 증가하는 정수를 꺼내 사용할수 있음
print(c)
```

```python
d = list(range(10, 0, -1)) # 10~1까지의 정수 중에서 1씩 감소하는 정수를 꺼내 사용할수 있음
print(d)
```
