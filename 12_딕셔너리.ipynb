```python
# 딕셔너리는 {} 중괄호내에 key:value이다.
# 딕셔너리 키 이름이 중복되면 마지막 key:value를 사용한다.
# 하나의 딕셔너리에 여러 자료형을 key:value로 사용할 수 있다.
# 딕셔너리의 키에 리스트와 딕셔너리는 사용할수 없다. tuple은 사용가능함(값 불변이기 떄문)
# dict로 딕셔너리를 만들수 있다.

# 딕셔너리 만들기
#딕셔너리 = {
  #  키1 : 값1, 
  #  키2 : 값2, ...
#}

lux = {
    "health" : 490,
    "mana" : 334,
    "melee" : 550,
    "armor" : 18.72
}

lux
```

```python
# 키 이름이 중복되면 마지막 키 : value를 사용함
lux = {
    "health" : 490,
    "health" : 800,
    "mana" : 334,
    "melee" : 550,
    "armor" : 18.72
}

lux
```

```python
# 딕셔너리의 키의 자료형
x = {
    100: "hundred",
    False : 0,
    3.5 : [3.5, 3.5]
}

x
```

```python
# 키에는 리스트와 딕셔너리를 사용할 수 없다. 
x = {
    [10, 20]: 100
}
x

x = {
    {10, 20}: 100
}
x
```

```python
x = {
    (10, 20): 100
}
x
```

```python
# 딕셔너리의 키에 접근하고 값 할당하기
lux = {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
print(lux['health'])
print(lux['armor'])
```

```python
# 딕셔너리의 키에 값 할당하기
lux = {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
lux['health']=500
lux['armor']=28.12
print(lux)
```
