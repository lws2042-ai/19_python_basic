```python
# 3의 배수와 5의 배수 출력 중복되는 공배수 처리가 안됨
for i in range(1, 101):
    
    if i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
for i in range(1, 101):
    
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
for i in range(1, 101):
    
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
for i in range(1, 101):
    print('Fizz' * (i % 3 == 0) + 'Buzz' * (i % 5 == 0) or i)
```

```python
# 20.8 심사문제
i, j = map(int, input().split())

if i > j:
    print("첫 번째 입력 값은 두 번째 입력 값보다 항상 작아야 합니다.")
else:
    for i in range(i, j+1):
        if i % 35 == 0:
            print(i,'FizzBuzz')
        else:
            print(i, 'Fizz' * (i % 5 == 0) + 'Buzz' * (i % 7 == 0) or i)
```
