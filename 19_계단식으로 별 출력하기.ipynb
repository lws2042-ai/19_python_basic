```python
# 중첩 루프 사용하기
for i in range(5):
    for j in range(5):
        print('j:', j, sep='', end=' ')
    print('i:', i, '\\n', sep='')
```

```python
# 중첩 루프
for i in range(5):
    for j in range(5):
        print('*', end='')
    print()
```

```python
for i in range(3):
    for j in range(7):
        print('*', end='')
    print()
```

```python
for i in range(5):
    for j in range(5):
        if j <= i:
            print('*', end='')
    print()
```

```python
for i in range(5):
    for j in range(5):
        if j == i:
            print('*', end='')
    print("")
```

```python
for i in range(5):
    for j in range(5):
        if j == i :
            print('*', end='')
        else:
            print(' ', end='')
    print()
```

```python
# 19.6 p219 심사문제

n = int(input())
space = list(reversed(range(n)))

for i in range(n):
    print(" " * space[i] + "*" * (2 * i + 1))
```
