```python
# 위치 인수와 리스트 언패킹 사용하기
def print_numbers(a, b,c ):
    print(a)
    print(b)
    print(c)
```

```python
print_numbers(10, 20, 30)
```

```python
# 언패킹 사용하기 
# 리스트나 튜플에 *(아스트리크)를 붙여 함수를 호출할 수 있다.
# 함수(*리스트)
# 함수(*튜플)
x = [10, 20, 30]
print_numbers(*x)
print_numbers(*[10, 20, 30])
```

```python
# 가변 인수 함수 만들기
def print_numbers(*args):
    for arg in args:
        print(arg)
```

```python
print_numbers(10)
print_numbers(10, 20, 30, 40)
```

```python
x = [10]
print_numbers(*x)

y = [10, 20, 30, 40, 50]
print_numbers(*y)
```

```python
# 키워드 인수 사용하기
def personal_info(name, age, address):
    print('이름: ', name)
    print('나이: ', age)
    print('주소: ', address)
```

```python
personal_info('홍길동', 30, '서울시 용산구 이촌동')
print()
personal_info(name = '홍길동', age = 30, address = '서울시 용산구 이촌동')
print()
personal_info(age = 30, address = '서울시 용산구 이촌동', name = '홍길동')
```

```python
# * 한개는 키, ** 두개는 값
personal_info(*{'name' : '홍길동', 'age' : 30, 'address' : '서울시 용산구 이촌동'})
print()
personal_info(**{'name' : '홍길동', 'age' : 30, 'address' : '서울시 용산구 이촌동'})
```

```python
x = {
    "name": "홍길동",
    "age": 30,
    "address": "서울시 용산구"

}
personal_info(*x)
print()
personal_info(**x)
```

```python
# 키워드 인수 함수 만들기
#def 함수이름(**매개변수):
#   코드

def personal_info(**kwargs):
    #print(kwargs)
    for kw, arg in kwargs.items():
        print(kw, arg)
```

```python
personal_info(name="홍길동", age=30, address="서울시 용산구")
```

```python
# 매개변수에 초기값 지정하기
def personal_info(name, age, address='비공개'):
    print('이름: ', name)
    print('나이: ', age)
    print('주소: ', address)
```

```python
personal_info("홍길동", 30)
print()
personal_info("홍길동", 30, "서울시 용산구")
```

```python
# 매개변수에 초기값 지정하기 default값이 지정될 인수는 제일뒤에 둬야한다.
def personal_info(name, address='비공개', age=0):
    print('이름: ', name)
    print('나이: ', age)
    print('주소: ', address)
```
